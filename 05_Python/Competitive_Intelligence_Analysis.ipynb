```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
ROOT=Path(r'E:/Data_Analysis/Dubai Automotive Intelligence Platform Sales, After-Sales & EV Adoption'); C=ROOT/'02_Data_Clean'
benchmark=pd.read_csv(C/'competitor_public_benchmark.csv'); score=pd.read_csv(C/'competitor_scorecard.csv')
plt.style.use('seaborn-v0_8-whitegrid'); display(benchmark); display(score)
```

## امتیاز کل رقبا

```python
x=score.sort_values('total_score'); x.plot.barh(x='company',y='total_score',legend=False,title='امتیاز کل'); plt.show()
print('تحلیل: '+'امتیاز ساختاریافته است و جایگزین سهم بازار حسابرسی\u200cشده نیست.')
```

## آمادگی EV

```python
x=score.sort_values('ev_readiness'); x.plot.barh(x='company',y='ev_readiness',legend=False,title='آمادگی EV'); plt.show()
print('تحلیل: '+'تعداد مدل EV کافی نیست و باید شارژ، تکنسین و قطعات هم اضافه شوند.')
```

## خدمات و فلیت

```python
score['service_fleet']=score.after_sales+score.fleet_mobility
x=score.sort_values('service_fleet'); x.plot.barh(x='company',y='service_fleet',legend=False,title='Q3 خدمات و فلیت'); plt.show()
print('تحلیل: '+'این ترکیب برای مشتریان B2B و حفظ مشتری مهم است.')
```

## تجربه دیجیتال

```python
x=score.sort_values('digital_experience'); x.plot.barh(x='company',y='digital_experience',legend=False,title='تجربه دیجیتال'); plt.show()
print('تحلیل: '+'مسیر مجازی تا تست\u200cدرایو و رزرو سرویس باید اندازه\u200cگیری شود.')
```

## پوشش برند

```python
x=score.sort_values('brand_count'); x.plot.barh(x='company',y='brand_count',legend=False,title='تعداد برندها'); plt.show()
print('تحلیل: '+'پوشش برند مزیت است اما پیچیدگی عملیات را بالا می\u200cبرد.')
```

## توانمندی خودروهای دست‌دوم

```python
x=score.sort_values('used_cars'); x.plot.barh(x='company',y='used_cars',legend=False,title='خودروهای دست\u200cدوم'); plt.show()
print('تحلیل: '+'بازرسی، گارانتی و تأمین مالی عامل اصلی رقابت هستند.')
```

## مراکز خدمات

```python
x=benchmark.dropna(subset=['service_center_count']).sort_values('service_center_count'); x.plot.barh(x='company',y='service_center_count',legend=False,title='Q7 مراکز خدمات'); plt.show()
print('تحلیل: '+'تعداد مرکز باید با ظرفیت و زمان انتظار ترکیب شود.')
```

## فلیت و موبیلیتی

```python
x=score.sort_values('fleet_mobility'); x.plot.barh(x='company',y='fleet_mobility',legend=False,title='فلیت و موبیلیتی'); plt.show()
print('تحلیل: '+'فلیت و اجاره برای تحلیل B2B اهمیت دارد.')
```

## پورتفولیو

```python
x=score.sort_values('portfolio_breadth'); x.plot.barh(x='company',y='portfolio_breadth',legend=False,title='پورتفولیو'); plt.show()
print('تحلیل: '+'تنوع سبد باید با سهم فروش و حاشیه سود سنجیده شود.')
```

## شکاف رقابتی

```python
score['score_vs_average']=score.total_score-score.total_score.mean()
x=score.sort_values('score_vs_average'); x.plot.barh(x='company',y='score_vs_average',legend=False,title='Q10 شکاف از میانگین'); plt.show()
print('تحلیل: '+'شکاف\u200cها باید با داده شعب و سهم بازار اعتبارسنجی شوند.')
```
